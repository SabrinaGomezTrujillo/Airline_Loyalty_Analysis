## 1 · Exploración y limpieza de datos

### 1.1 EDA inicial


```python
def resumen_eda(df: pd.DataFrame, nombre: str, top_n: int = 3) -> None:
    """EDA rápido: shape, duplicados, nulos, descriptiva numérica y categórica."""
    sep = '=' * 60
    print(f'{sep}\nEDA · {nombre.upper()}\n{sep}')
    print(f'Shape: {df.shape}')
    dup = df.duplicated().sum()
    print(f'Duplicados: {dup:,} ({100*dup/len(df):.2f}%)\n')

    nulos_pct = (100 * df.isnull().sum() / len(df))
    nulos_pct = nulos_pct[nulos_pct > 0]
    if not nulos_pct.empty:
        print('Nulos (%):\n', nulos_pct.round(2).to_string(), '\n')

    print('Descriptiva numérica:')
    display(df.describe().round(2))

    cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns
    if len(cat_cols):
        print(f'\nDescriptiva categórica ({len(cat_cols)} variables):')
        for col in cat_cols:
            n_u = df[col].nunique()
            top = ', '.join(f"'{v}'({c})" for v, c in df[col].value_counts().head(top_n).items())
            print(f'  {col}: {n_u} categorías — top {min(top_n,n_u)}: {top}')
```

```python
resumen_eda(flight_df,  'Customer Flight Activity')
print()
resumen_eda(loyalty_df, 'Customer Loyalty History')
```

### 1.2 Limpieza de datos


#### 1.2.1 `flight_df` — Eliminación de duplicados

Se detectaron registros distintos para el mismo `(Loyalty Number, Year, Month)`,
lo que es inconsistente si `Flights Booked` y `Distance` representan acumulados mensuales.
**Decisión:** conservar el primer registro por combinación clave; se eliminan 3 936 filas (0.97%).


```python
shape_0 = flight_df.shape
flight_df = flight_df.drop_duplicates(subset=['Loyalty Number', 'Year', 'Month'], keep='first')
eliminadas = shape_0[0] - flight_df.shape[0]
print(f'Filas eliminadas : {eliminadas:,} ({100*eliminadas/shape_0[0]:.2f}%)')
print(f'Shape final      : {flight_df.shape}')
```

#### ⚠️ Nota sobre duplicados: diferencia entre el EDA y la limpieza final

Durante el EDA se detectaron **1.864 filas duplicadas exactas** (0.46%) —
registros donde todas las columnas tienen valores idénticos.

Sin embargo, al agrupar por `(Loyalty Number, Year, Month)` se identificaron
**3.936 filas** (0.97%) incluyendo las filas duplicadas exactas y registros que no son duplicados exactos pero sí son
**inconsistencias de datos**: un mismo cliente tiene más de un registro para el
mismo período con valores distintos de `Flights Booked`, `Distance` u otras
variables que deberían ser únicas por mes.

##### Ejemplo observado

Al inspeccionar el cliente `Loyalty Number = 101902` se encontró que para un
mismo mes la distancia registrada se *reduce* entre registros, lo cual es
imposible si `Distance` representa un acumulado mensual. Esto confirma que los
registros adicionales no son observaciones válidas sino errores en la fuente.

| Tipo de duplicado | Filas | % | Detección |
|---|---|---|---|
| Duplicados exactos (todas las columnas iguales) | 1.864 | 0.46% | `df.duplicated()` en el EDA |
| Inconsistencias por clave `(Loyalty Number, Year, Month)` | 3.936 | 0.97% | `drop_duplicates(subset=[...])` |


> **Decisión:** se conservó únicamente el **primer registro** por combinación
> `(Loyalty Number, Year, Month)`, eliminando tanto los duplicados exactos como
> las inconsistencias. La pérdida de información es mínima (< 1%) y está
> justificada por la falta de fiabilidad de los registros adicionales.
> El dataset pasa de **405.624 a 401.688 registros**.

#### 1.2.2 `loyalty_df` — Salarios, cancelaciones y columna constante

| Paso | Acción | Resultado |
|------|--------|-----------|
| a) `Country` constante | `drop('Country')` | -1 columna |
| b) Salarios negativos | Convertir a NaN | 20 valores corregidos |
| c) Nulos en `Salary` (25.45%) | Imputar mediana por provincia | 0 nulos restantes |
| d) `Cancellation Year/Month` (87.7% nulos) | Crear flag `Cancelled` | 2 067 cancelados (12.35%) |


```python
# a) Eliminar columna constante
loyalty_df.drop('Country', axis=1, inplace=True)

# b) Salarios negativos -> NaN
neg = (loyalty_df['Salary'] < 0).sum()
loyalty_df.loc[loyalty_df['Salary'] < 0, 'Salary'] = None
print(f'Salarios negativos convertidos a NaN: {neg}')

# c) Imputar con mediana por provincia
loyalty_df['Salary'] = loyalty_df.groupby('Province')['Salary'].transform(
    lambda x: x.fillna(x.median())
)
print(f'Nulos restantes en Salary: {loyalty_df["Salary"].isnull().sum()}')

# d) Flag de cancelacion
loyalty_df['Cancelled'] = loyalty_df['Cancellation Year'].notna()
print(f'Clientes cancelados: {loyalty_df["Cancelled"].sum():,} '
      f'({100*loyalty_df["Cancelled"].mean():.2f}%)')
```

### 1.3 Unión de datasets

Se realiza un **LEFT JOIN** de `flight_df` (actividad mensual) con `loyalty_df`
(atributos estáticos del cliente) sobre `Loyalty Number`, preservando las
401 688 observaciones de actividad y enriqueciéndolas con el perfil del socio.


```python
flight_df['Loyalty Number']  = flight_df['Loyalty Number'].astype(int)
loyalty_df['Loyalty Number'] = loyalty_df['Loyalty Number'].astype(int)

df = flight_df.merge(loyalty_df, on='Loyalty Number', how='left')

print(f'Shape combinado  : {df.shape}')
print(f'Sin match en join: {df["Province"].isnull().sum()}')
df.head()
```

## 2 · Análisis estadístico

### 2.1 Variables numéricas

#### a) Estadística descriptiva


```python
excluir  = ['Cancellation Year', 'Cancellation Month']
num_cols = [c for c in df.select_dtypes(include=['int64', 'float64']).columns
            if c not in excluir]

resumen = []
for col in num_cols:
    s = df[col]
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    resumen.append({
        'Variable'  : col,
        'Media'     : round(s.mean(), 2),
        'Mediana'   : round(s.median(), 2),
        'Desv. Std' : round(s.std(), 2),
        'Min'       : round(s.min(), 2),
        'Max'       : round(s.max(), 2),
        'Q1'        : round(q1, 2),
        'Q3'        : round(q3, 2),
        'IQR'       : round(q3 - q1, 2),
        '% Nulos'   : round(100 * s.isnull().mean(), 2),
    })

display(pd.DataFrame(resumen).set_index('Variable'))
```

> **Interpretación clave**
> - **Asimetría positiva** en toda la actividad: la mayoría de socios tiene valores bajos y una cola larga de alto volumen.
> - **75% nunca canjea puntos** (`Points Redeemed = 0`) — el programa de puntos está infrautilizado; oportunidad directa de mejora.
> - **CLV**: mediana 5 780 vs media 7 989 — existe un segmento de clientes de altísimo valor.
> - **Salary**: la mediana (73 455) es más representativa que la media (79 246), afectada por outliers altos.


#### b) Detección de outliers (método IQR)


```python
def outliers_iqr(df, col):
    q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    iqr = q3 - q1
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    mask = (df[col] < lo) | (df[col] > hi)
    return mask.sum(), round(100 * mask.mean(), 2), round(lo, 2), round(hi, 2)

rows = []
for col in num_cols:
    n, pct, lo, hi = outliers_iqr(df, col)
    rows.append({'Variable': col, 'Outliers': n, '% Outliers': pct,
                 'Limite inf.': lo, 'Limite sup.': hi,
                 'Max real': round(df[col].max(), 2),
                 'Min real': round(df[col].min(), 2)})

display(pd.DataFrame(rows).set_index('Variable')
          .sort_values('% Outliers', ascending=False))
```

> **Decision:** los outliers reflejan **comportamientos reales** (viajeros frecuentes, clientes de alto CLV,
> ingresos premium), no errores de datos. Se **mantienen** en el dataset.
> En modelos futuros se aplicaran transformaciones logaritmicas o winsorizacion si fuera necesario.

| Variable | % Outliers | Interpretacion |
|----------|-----------|----------------|
| `Flights with Companions` | 17.7% | Segmento familiar/social con alta frecuencia |
| `CLV` | 8.9% | Clientes de altisimo valor estrategico |
| `Points Redeemed` | 6.0% | Canjeadores activos (minoria significativa) |
| `Salary` | 5.1% | Segmento de alto poder adquisitivo |


#### c) Correlaciones


```python
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(12, 10))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.4, ax=ax)
ax.set_title('Matriz de correlaciones', fontsize=14, pad=12)
plt.tight_layout()
plt.show()

# Pares con correlacion > 0.7
high = [(corr.columns[i], corr.columns[j], round(corr.iloc[i,j], 4))
        for i in range(len(corr)) for j in range(i+1, len(corr))
        if abs(corr.iloc[i,j]) > 0.7]
display(pd.DataFrame(high, columns=['Var 1','Var 2','r'])
          .sort_values('r', ascending=False).reset_index(drop=True))
```

> **Correlaciones clave**
> - `Distance` <-> `Points Accumulated`: r = 0.99 — los puntos reflejan fielmente la distancia volada.
> - `Points Redeemed` <-> `Dollar Cost Points Redeemed`: r > 0.99 — variables redundantes; conservar solo una.
> - `Flights Booked` <-> `Total Flights`: r = 0.96 — `Total Flights` es derivable; puede eliminarse en modelos.
> **Reduccion de dimensionalidad recomendada:** eliminar 2-3 variables redundantes sin perdida de informacion.

## 3 · Visualizaciones


### 3.1 Vuelos reservados por mes (estacionalidad)

**Objetivo:** identificar patrones de demanda a lo largo del año y diferencias entre 2017 y 2018.


```python
monthly = df.groupby(['Year', 'Month'])['Flights Booked'].agg(['sum', 'mean']).reset_index()
meses = ['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for year, grp in monthly.groupby('Year'):
    ax1.plot(grp['Month'], grp['sum'],  marker='o', label=str(year))
    ax2.plot(grp['Month'], grp['mean'], marker='o', label=str(year))

for ax, title, ylabel in [
    (ax1, 'Total de vuelos reservados por mes', 'Vuelos (total)'),
    (ax2, 'Promedio de vuelos por cliente',     'Vuelos (promedio)'),
]:
    ax.set_xticks(range(1, 13))
    ax.set_xticklabels(meses)
    ax.set_title(title)
    ax.set_xlabel('Mes')
    ax.set_ylabel(ylabel)
    ax.legend(title='Año')
    ax.grid(True, linestyle='--', alpha=0.4)

plt.suptitle('Estacionalidad de reservas — 2017 vs 2018', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

> **Insight:** hay una **estacionalidad clara** con picos en verano (jun–ago) y diciembre.
> En 2018 el volumen crece a partir de mayo, tanto en total como en promedio por cliente.
> Los meses bajos (ene–abr) son la ventana ideal para **promociones de estimulacion de demanda**.


### 3.2 Distancia volada vs puntos acumulados

**Objetivo:** verificar la consistencia del programa de puntos y detectar cambios de politica entre años.


```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, year in zip(axes, [2017, 2018]):
    sub = df[df['Year'] == year].sample(5000, random_state=42)
    ax.scatter(sub['Distance'], sub['Points Accumulated'], alpha=0.3, s=8)
    m, b = np.polyfit(sub['Distance'], sub['Points Accumulated'], 1)
    x_line = np.linspace(sub['Distance'].min(), sub['Distance'].max(), 100)
    ax.plot(x_line, m * x_line + b, color='red', lw=1.5)
    r = sub['Distance'].corr(sub['Points Accumulated'])
    ax.set_title(f'{year}  |  r = {r:.4f}  |  pendiente = {m:.2f} pts/km')
    ax.set_xlabel('Distancia (km)')
    ax.set_ylabel('Puntos acumulados')

fig.suptitle('Relacion distancia vs puntos acumulados por año', fontsize=13)
plt.tight_layout()
plt.show()
```

> **Insight:** correlacion practicamente perfecta (r aprox 0.99) en ambos años — valida la calidad del dataset.
> La **pendiente aumento en 2018**, indicando un cambio de politica: se otorgan mas puntos por km volado.
> Al modelar puntos acumulados, `Year` debe incluirse como variable de control.

## 4 · Prueba de hipotesis: vuelos reservados por nivel educativo

**Pregunta:** ¿El numero total de vuelos reservados, difiere significativamente segun el nivel educativo del cliente?

- **H0:** las medias son iguales en todos los niveles educativos.
- **H1:** al menos un nivel educativo difiere significativamente.

Se aplican ANOVA (parametrico) y Kruskal-Wallis (no parametrico) como pruebas complementarias.


```python
# Vuelos totales por cliente
fpc = (df.groupby(['Loyalty Number', 'Education'])['Flights Booked']
         .sum().reset_index()
         .rename(columns={'Flights Booked': 'Total_Vuelos'}))

# Descriptiva por nivel educativo
print('Estadisticas de vuelos totales por nivel educativo:')
display(fpc.groupby('Education')['Total_Vuelos'].describe().round(2))

# Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(data=fpc, x='Education', y='Total_Vuelos', palette='Set2')
plt.title('Vuelos totales reservados por nivel educativo')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# Pruebas estadisticas
groups    = [g['Total_Vuelos'].values for _, g in fpc.groupby('Education')]
f_stat, p_anova = f_oneway(*groups)
h_stat, p_kw    = kruskal(*groups)

print(f'ANOVA          F = {f_stat:.4f} | p = {p_anova:.4f}')
print(f'Kruskal-Wallis H = {h_stat:.4f} | p = {p_kw:.4f}')
```

### Resultados

| Prueba | Estadistico | p-valor | Conclusion |
|--------|-------------|---------|------------|
| ANOVA | F = 0.8564 | 0.4893 | No significativo |
| Kruskal-Wallis | H = 3.6735 | 0.4520 | No significativo |

> **Conclusion:** con p >> 0.05 en ambas pruebas, **no existe evidencia estadistica** de que el numero de vuelos
> difiera por nivel educativo. Las medias observadas son practicamente identicas (98.7–101.0 vuelos por cliente).
> `Education` no es un driver de actividad de vuelo.
>
> **Implicacion practica:** las campañas orientadas a incrementar reservas pueden diseñarse sin segmentar
> por educacion. Variables con mayor poder discriminante son `Loyalty Card`, `Province` y `CLV`.
